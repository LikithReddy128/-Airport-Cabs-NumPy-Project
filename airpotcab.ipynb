```python
import numpy as np
```

```python
taxi = np.genfromtxt('nyc_taxis.csv', delimiter=',', skip_header=True)
```

```python
speed = taxi[:, 7] / (taxi[:, 8] / 3600)
```

```python
mean_speed = speed.mean()
print(mean_speed)
```
